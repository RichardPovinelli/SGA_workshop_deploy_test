```python
import importlib.util
import os
import subprocess
import sys
from pathlib import Path


def ensure_repository_available(
    repo_url: str,
    repo_dir: str,
) -> None:
    """Clone or update repository if src/ directory doesn't exist."""
    if not Path("src").exists():
        if not Path(repo_dir).exists():
            subprocess.check_call(["git", "clone", repo_url, "-q"])
        else:
            subprocess.check_call(["git", "-C", repo_dir, "pull", "-q"])
        os.chdir(repo_dir)


def ensure_dependencies_installed() -> None:
    """Install required packages if pandas or xgboost are missing."""
    if importlib.util.find_spec("pandas") is None or importlib.util.find_spec("xgboost") is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", "requirements.txt", "-q"])


def ensure_repo_in_path() -> None:
    """Add repository root to sys.path if not already present."""
    repo_root = Path.cwd()
    if str(repo_root) not in sys.path:
        sys.path.insert(0, str(repo_root))


def setup_environment(
    repo_url: str = "https://github.com/RichardPovinelli/SGA_workshop_deploy_test.git",
    repo_dir: str = "SGA_workshop_deploy_test",
) -> None:
    """Set up the workshop environment.
    
    Ensures repository is available, dependencies are installed,
    and repo root is in sys.path.
    """
    ensure_repository_available(repo_url, repo_dir)
    ensure_dependencies_installed()
    ensure_repo_in_path()


setup_environment()
```

## Part 2: Baseline Models

Run the baseline models for one selected horizon and compare metrics.

```python
# Train baseline models for horizon 1 (for pairwise comparison)
from src.data.split import get_test_df
from src.models.elasticnet import fit_elasticnet_model, predict_elasticnet
from src.models.lasso import fit_lasso_model, predict_lasso
from src.models.linear import fit_linear_model, predict_linear
from src.models.naive import predict_naive

test_df = get_test_df(df)
horizon = 1

# Train and predict for horizon 1
naive_predictions = predict_naive(test_df, horizon)

linear_model = fit_linear_model(df, horizon)
linear_predictions = predict_linear(linear_model, test_df)

lasso_model = fit_lasso_model(df, horizon)
lasso_predictions = predict_lasso(lasso_model, test_df)

elasticnet_model = fit_elasticnet_model(df, horizon)
elasticnet_predictions = predict_elasticnet(elasticnet_model, test_df)

print("Baseline models trained for horizon 1")
```

```python
# Compare all baseline models pairwise using one-sided Diebold-Mariano test
import pandas as pd
from src.evaluation.dm_test import diebold_mariano_test


models = {
    "naive": naive_predictions,
    "linear": linear_predictions,
    "lasso": lasso_predictions,
    "elasticnet": elasticnet_predictions,
}

target = test_df["target_h1"]
model_names = list(models.keys())

# Create matrix of p-values (one-sided test: row model better than column)
n_models = len(model_names)
pvalue_matrix = pd.DataFrame(index=model_names, columns=model_names, dtype=float)

for i, m1 in enumerate(model_names):
    for j, m2 in enumerate(model_names):
        if i == j:
            pvalue_matrix.loc[m1, m2] = 1.0
        else:
            result = diebold_mariano_test(target, models[m1], models[m2], alternative="greater")
            pvalue_matrix.loc[m1, m2] = result["p_value"]

print("Diebold-Mariano Test (One-Sided): p-values (rows vs columns)")
print("(p < 0.05 means row model is significantly better than column model)")
print(pvalue_matrix.astype(float).round(2))
```

```python
import pandas as pd
from math import log10, floor

from src.data.split import get_test_df
from src.models.elasticnet import evaluate_elasticnet_model, predict_elasticnet
from src.models.lasso import evaluate_lasso_model, predict_lasso
from src.models.linear import evaluate_linear_model, predict_linear
from src.models.naive import evaluate_naive_model, predict_naive

test_df = get_test_df(df)

# Train all baseline models across all 7 horizons
all_baseline_results = []
for h in range(1, 3):
    naive_results = evaluate_naive_model(test_df, h)
    _, linear_results = evaluate_linear_model(df, h)
    _, lasso_results = evaluate_lasso_model(df, h)
    _, elasticnet_results = evaluate_elasticnet_model(df, h)
    
    all_baseline_results.extend([naive_results, linear_results, lasso_results, elasticnet_results])

comparison = pd.concat(all_baseline_results, ignore_index=True)

# Round to 2 significant digits
def round_sig(x, sig=3):
    if x == 0:
        return 0
    return round(x, -int(floor(log10(abs(x)))) + (sig - 1))

comparison[['mape', 'wmape', 'rmse']] = comparison[['mape', 'wmape', 'rmse']].apply(
    lambda col: col.apply(lambda x: round_sig(x, 3))
)
comparison
```

```python
# Compare baseline models visually
from src.visualization.plots import plot_model_metrics_comparison
plot_model_metrics_comparison(comparison, "rmse", "Baseline Model Comparison (RMSE)")
```

## Part 3: XGBoost

Train the workshop XGBoost model and compare it to the linear baselines.

```python
# Train XGBoost and Linear across horizons 1-7 for multi-horizon comparison
from src.models.xgb import evaluate_xgb

all_xgb_linear_results = []
for h in range(1, 8):
    xgb_res = evaluate_xgb(df, h)
    _, xgb_results = xgb_res
    _, linear_results = evaluate_linear_model(df, h)
    
    all_xgb_linear_results.extend([xgb_results, linear_results])

xgb_linear_comparison = pd.concat(all_xgb_linear_results, ignore_index=True)

xgb_linear_comparison[['mape', 'wmape', 'rmse']] = xgb_linear_comparison[['mape', 'wmape', 'rmse']].apply(
    lambda col: col.apply(lambda x: round_sig(x, 3))
)

# Display only horizons 1-2
xgb_linear_comparison[['horizon', 'model', 'mape', 'wmape', 'rmse']].loc[xgb_linear_comparison['horizon'] <= 2]
```

```python
# Grouped bar chart: XGBoost vs Linear RMSE across all horizons
import plotly.graph_objects as go

fig = go.Figure()

models = sorted(xgb_linear_comparison["model"].unique())
horizons = sorted(xgb_linear_comparison["horizon"].unique())

# Define colors for each model
colors = {
    "xgb": "#ff7f0e",      # orange
    "linear": "#1f77b4",   # blue
}

# Add bars for each model
for model in models:
    model_data = xgb_linear_comparison[xgb_linear_comparison["model"] == model].sort_values("horizon")
    
    fig.add_trace(go.Bar(
        x=model_data["horizon"],
        y=model_data["rmse"],
        name=model.capitalize(),
        marker=dict(color=colors.get(model, "#000000")),
    ))

fig.update_layout(
    title="XGBoost vs Linear: RMSE Across All Horizons",
    xaxis_title="Forecast Horizon",
    yaxis_title="RMSE",
    barmode="group",
    xaxis=dict(dtick=1),
    hovermode="x unified",
    template="plotly_white",
    height=500,
    showlegend=True,
)
fig.show()
```

```python
# One-sided Diebold-Mariano test: XGBoost vs Linear (is XGBoost better?)
# Test each horizon separately
from src.models.xgb import fit_xgb, predict_xgb

for h in range(1, 3):
    # Train models
    xgb_model_temp = fit_xgb(df, h)
    linear_model_temp = fit_linear_model(df, h)
    
    # Get test data
    test_df_h = get_test_df(df)
    
    # Get predictions
    xgb_pred = predict_xgb(xgb_model_temp, test_df_h)
    linear_pred = predict_linear(linear_model_temp, test_df_h)
    
    # Get target
    target_col = f"target_h{h}"
    target = test_df_h[target_col]
    
    # Run DM test (XGBoost vs Linear, is XGBoost better?)
    result = diebold_mariano_test(target, xgb_pred, linear_pred, alternative="greater")
    
    print(f"Horizon {h}: XGBoost vs Linear")
    print(f"  p-value (XGBoost better): {result['p_value']:.4f}")
    print(f"  Significant at 0.05? {result['p_value'] < 0.05}\n")
```
