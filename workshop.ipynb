```python
import importlib.util
import os
import shutil
import subprocess
import sys
from pathlib import Path


def is_colab() -> bool:
    """Check if running on Google Colab."""
    return "COLAB_RELEASE_TAG" in os.environ


def clean_repo_directory(repo_dir: str) -> None:
    """Delete repository directory if it exists.
    
    Args:
        repo_dir: Local directory name for the repository.
    """
    repo_path = Path(repo_dir)
    if repo_path.exists():
        shutil.rmtree(repo_path)


def ensure_repository_available(
    repo_url: str,
    repo_dir: str,
) -> None:
    """Clone or update repository if src/ directory doesn't exist.
    
    On Colab: Always performs a fresh clone after deleting any existing repo.
    On local: Clones if missing, pulls if already exists.
    
    Args:
        repo_url: URL of the git repository to clone.
        repo_dir: Local directory name for the repository.
    """
    # On Colab: always do a fresh clone
    if is_colab():
        clean_repo_directory(repo_dir)
        subprocess.check_call(["git", "clone", repo_url, "-q"])
        os.chdir(repo_dir)
    # On local: keep existing pull/clone logic
    else:
        if not Path("src").exists():
            if not Path(repo_dir).exists():
                subprocess.check_call(["git", "clone", repo_url, "-q"])
            else:
                subprocess.check_call(["git", "-C", repo_dir, "pull", "-q"])
            os.chdir(repo_dir)


def ensure_dependencies_installed() -> None:
    """Always install required packages, beacause a package may not exist on Colab."""
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", "requirements.txt", "-q"])


def ensure_repo_in_path() -> None:
    """Add repository root to sys.path if not already present."""
    repo_root = Path.cwd()
    if str(repo_root) not in sys.path:
        sys.path.insert(0, str(repo_root))


def setup_environment(
    repo_url: str = "https://github.com/RichardPovinelli/SGA_workshop_deploy.git",
    repo_dir: str = "SGA_workshop_deploy",
) -> None:
    """Set up the workshop environment.
    
    Ensures repository is available, dependencies are installed,
    and repo root is in sys.path.
    """
    ensure_repository_available(repo_url, repo_dir)
    ensure_dependencies_installed()
    ensure_repo_in_path()


setup_environment()
```

## Part 2: Baseline Models

Run the baseline models for one selected horizon and compare metrics.

```python
# Train baseline models for horizon 1 (for pairwise comparison)
from src.data.split import get_test_df
from src.models.elasticnet import fit_elasticnet_model, predict_elasticnet
from src.models.lasso import fit_lasso_model, predict_lasso
from src.models.linear import fit_linear_model, predict_linear
from src.models.naive import predict_naive

test_df = get_test_df(df)
horizon = 1

# Train and predict for horizon 1
naive_predictions = predict_naive(test_df, horizon)

linear_model = fit_linear_model(df, horizon)
linear_predictions = predict_linear(linear_model, test_df)

lasso_model = fit_lasso_model(df, horizon)
lasso_predictions = predict_lasso(lasso_model, test_df)

elasticnet_model = fit_elasticnet_model(df, horizon)
elasticnet_predictions = predict_elasticnet(elasticnet_model, test_df)

print("Baseline models trained for horizon 1")
```

```python
# Compare baseline models visually - all horizons
import pandas as pd
from src.visualization.plots import plot_model_metrics_comparison
from src.evaluation.metrics import build_results_table
from src.models.elasticnet import fit_elasticnet_model, predict_elasticnet
from src.models.lasso import fit_lasso_model, predict_lasso
from src.models.linear import fit_linear_model, predict_linear
from src.models.naive import predict_naive

# Train and evaluate all baseline models for all horizons
comparison_data = []
for h in range(1, 8):
    target_col = f"target_h{h}"
    target = test_df[target_col]
    
    # Get predictions for this horizon
    naive_pred = predict_naive(test_df, h)
    
    linear_model = fit_linear_model(df, h)
    linear_pred = predict_linear(linear_model, test_df)
    
    lasso_model = fit_lasso_model(df, h)
    lasso_pred = predict_lasso(lasso_model, test_df)
    
    elasticnet_model = fit_elasticnet_model(df, h)
    elasticnet_pred = predict_elasticnet(elasticnet_model, test_df)
    
    # Build results for each model
    comparison_data.append(build_results_table(target, naive_pred, model_name="naive", horizon=h))
    comparison_data.append(build_results_table(target, linear_pred, model_name="linear", horizon=h))
    comparison_data.append(build_results_table(target, lasso_pred, model_name="lasso", horizon=h))
    comparison_data.append(build_results_table(target, elasticnet_pred, model_name="elasticnet", horizon=h))

comparison = pd.concat(comparison_data, ignore_index=True)

plot_model_metrics_comparison(comparison, "rmse", "Baseline Model Comparison (RMSE)")
```

### How To Read This DM Table

- Each cell is a one-sided Diebold-Mariano p-value comparing the row model to the column model.
- Null hypothesis: the row model is not better than the column model.
- Smaller p-values mean stronger evidence that the row model has lower forecast loss.
- A common threshold is $p < 0.05$ for statistical significance.
- Diagonal values are 1.0 because a model compared with itself has no difference.

```python
# Compare all baseline models pairwise using one-sided Diebold-Mariano test
from src.evaluation.comparisons import evaluate_dm_table_from_predictions

predictions = {
    "naive": naive_predictions,
    "linear": linear_predictions,
    "lasso": lasso_predictions,
    "elasticnet": elasticnet_predictions,
}
target = test_df["target_h1"]
model_names = list(predictions.keys())

dm_matrix_display = evaluate_dm_table_from_predictions(
    predictions,
    target,
    model_names,
    sig=2,
 )

print("Diebold-Mariano Test (One-Sided): p-values (rows vs columns)")
print("(p < 0.05 means row model is significantly better than column model)")
dm_matrix_display
```

```python
# One-sided Diebold-Mariano test: XGBoost vs Linear (is XGBoost better?)
# Test each horizon separately
from src.models.xgb import fit_xgb, predict_xgb
from src.evaluation.dm_test import diebold_mariano_test

for h in range(1, 3):
    # Train models
    xgb_model_temp = fit_xgb(df, h)
    linear_model_temp = fit_linear_model(df, h)
    
    # Get test data
    test_df_h = get_test_df(df)
    
    # Get predictions
    xgb_pred = predict_xgb(xgb_model_temp, test_df_h)
    linear_pred = predict_linear(linear_model_temp, test_df_h)
    
    # Get target
    target_col = f"target_h{h}"
    target = test_df_h[target_col]
    
    # Run DM test (XGBoost vs Linear, is XGBoost better?)
    result = diebold_mariano_test(target, xgb_pred, linear_pred, alternative="greater")
    
    print(f"Horizon {h}: XGBoost vs Linear")
    print(f"  p-value (XGBoost better): {result['p_value']:.4f}")
    print(f"  Significant at 0.05? {result['p_value'] < 0.05}\n")
```
